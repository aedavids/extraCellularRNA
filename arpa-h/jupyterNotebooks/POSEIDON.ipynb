Daniel slack D.M. 8/5/24 "Is there a pan cancer signal that you see in all TCGA but no GTEx?"

Run the analysis twice once at the gene level the other at the transcript level

## Design

1) split normalized count data  into a GTEx dataframe and a TCGA DataFrame.  
    see ciber sort code for normalization
2) for each data set create a logical data frame. count > 0.0
3) calculate the row (gene or transcript) mean of the logical.
   If mean is 1, all samples express that gene. If zero no samples express genes. Values in between are a measure of how many sample express the gene
4) Create A results Data Frame with following columns  
   [nGTEx, nTCGA, GTExMean, GTExMean - TCGAMean, GTExMean - (GTExMean - TCGAMean)]
   - number of sample that have non zero count
   - GTExMean range [0,1]
   - GTExMean - TCGAMean range = [-1,1].
       + negitive values means high percentage of TCGA samples express transcript/count  
       + zero mean expressed in both data sets  
       + positive value close to 1 means only expressed in GTEx  
       + negative value close to -1 means only expressed in TCGA  
   - GTExMean - (GTExMean - TCGAMean)
       + means may not be close to zero.
       + near zero means TCGAMean was near zero zero. potential good biomarker not all samples have it but only GTEx.
       + negative means strong TCGA
